```python
import os
import scanpy as sc
import pandas as pd

# --- Configuration ---
base_out_dir = "/data/ag_reichart_lindberg/nhadizad/nastaran/projects/xeno/analysis/v1/CLR/v1/cell_state/ref_3month"
target_cell_types = ["CM", "FB", "EC", "Mye_papio", "Lymph_papio", "Mye_sus", "Lymph_sus"]

for ct in target_cell_types:
    print(f"Processing {ct}...")
    
    # Create the specific subfolder for this cell type
    ct_dir = os.path.join(base_out_dir, ct)
    os.makedirs(ct_dir, exist_ok=True)
    
    # Subset the AnnData object for the current cell type
    adata_subset = adata[adata.obs['cell_type'] == ct].copy()
    
    # Extract only the necessary columns (Sample_ID, cell_state, Condition)
    metadata_df = adata_subset.obs[['Sample_ID', 'cell_type', 'cell_state', 'Condition']].copy()
    
    # Drop rows where cell_state might be NaN (optional but good practice)
    metadata_df = metadata_df.dropna(subset=['cell_state'])
    
    # Save to CSV inside the subfolder (index=False prevents the unnamed column error in R)
    csv_path = os.path.join(ct_dir, "metadata_for_clr_cellstate.csv")
    metadata_df.to_csv(csv_path, index=False)
    
print(f"All metadata successfully saved to {base_out_dir}")
```
